**args and kwargs**

- args means arguments we can pass multiple arguments at a time
  - output come in tuple

- kwrgs means keywords arguments we can mutiple arguments at time
  - output come in a dictionary

```python
a=10
b=5
a+b
```

```python
def f1(*args):
    print(args)
f1(10,20,30)
```

```python
def f1(**kwargs):
    print(kwargs)
f1(a=10,b=20,c=30)
```

```python
def f1(**kwargs):
    print(kwargs)
f1(name= 'Pankaj',age = '21',city='Hyd')
```

```python
def Math():
    a=10
    b=20
    add=a+b # 30
    sub=a-b # -10
    return(add)

add=Math()
```

```python
def Math():
    a=10
    b=20
    add=a+b # 30
    sub=a-b # -10
    return(add,sub)

add=Math()
```

```python
def f1(n):
    print(n)
    if n==0:
        return(1)
    return  n*f1(n-1)

ans=f1(5)
print(ans)
# 5 == 5 == 5-1  ==4*5=20
# 4 == 4 == 4-1 == 3*20= 60
# 3 == 3 == 3-1 == 2*60 120
# 2 == 2 == 2-1 = 1*120 = 120
# 1 == 1 == 1-1 = 0* 120 = 0
```
